# Using list vs arrays for mathematical calculations

```python
a = [11, 15, 13, 12]
print(a, type(a))
```

```python
a * 3
```

```python
# Multiply 3 to every number in list
b = []

for i in a:
    b.append(i * 3)

# Print the b list
print(f"{a} x 3 = {b}")
```

```python
a1 = np.array(a)
print(a1, type(a1))
```

```python
# On array for loop is not required, you can perform broadcasting
b1 = a1 * 3
print(b1)
```

```python
a1
```

```python
a1 + 5
```

```python
5 * a1 + 4
```

```python
a1**2
```

```python
a1**3
```

```python
a1 ** (1 / 2)
```

# Random normal distribution
