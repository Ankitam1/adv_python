# Generic Exception handling

```python
help(Exception)
```

```python
def division2(num: int | float, den: int | float) -> float | None:
    try:
        div = num / den
        return div
    except Exception as e:
        print(f"Exception occured : {e}")
```

```python
a = division2(num=100, den=26)
print(a)
print("Completed")
```

```python
b = division2(num=25, den=0)
print(b)
print("Completed")
```

```python
c = division2(num="abc", den=4)
print(c)
print("Completed")
```

```python
d = division2(num="a", den="b")
print(d)
print("Completed")
```

# Use differnent exceptions for each errors

```python
def division3(num: int | float, den: int | float) -> float | None:
    try:
        div = num / den
        return div
    except ZeroDivisionError:
        print("Denomator of division cannot be Zero")
    except TypeError:
        print("Provide number input for numerator and denominator")
    except Exception as e:
        print(f"Unexpcted exception occured : {e}")
```

```python
a = division3(num=23, den=6)
print(a)
print("Complete")
```

```python
b = division3(num=35, den=0)
print(b)
print("Complete")
```

```python
c = division3(num=3, den="b")
print(c)
print("Complete")
```

# You can also raise an error in python whenever required


```python
def age_validator(age: int) -> int | None:
    if age <= 0 or age >= 100:
        raise ValueError("Age should be between 1 and 100")
    return age
```

```python
a1 = age_validator(25)
print(a1)
```

```python
a2 = age_validator(-23)
print(a2)
```

```python
a3 = age_validator("abc")
```

# Other blocks
1. try - Code you want to run
2. except - If exception occurs in try block , exception block will be triggered
3. else - else block is executed only if try block is successful
4. finally - finally block is always executed in the end

```python
import math

math.sqrt(23)
```

```python
math.sqrt(4)
```

```python
math.sqrt(-1)
```

```python
math.sqrt("a")
```

```python
def square_root(num: int | float) -> float | None:
    try:
        s = math.sqrt(num)
    except Exception as e:
        print(f"Exception occured : {e}")
    else:
        print(f"Square root of number {num} is {s:.4f}")
        return s
    finally:
        print("Finally block is always executed")
```

```python
s1 = square_root(num=16)
```

```python
s1
```

```python
s2 = square_root(num=12)
```

```python
s2
```

```python
s3 = square_root(num=-15)
```

```python
print(s3)
```

```python
s5 = square_root(num="abc")
```

```python
s6 = square_root([1, 2, 3, 4])
```

```python
print(s6)
```

# File Handling
1. read a file - 'r'
2. write a new file - 'w'
3. append new line in a file - 'a'

```python
# Writing a completely new file
with open("new.txt", "w", encoding="utf-8") as f:
    f.write("My name is Ankita Mirkale \n")
    f.write("I work as a Senior CP \n")
```

```python
# Reading the file
with open("new.txt", "r", encoding="utf-8") as f:
    content = f.read()

# Print the content
print(content)
```

```python
# Add contect in file append mode
with open("new.txt", "a", encoding="utf-8") as f:
    f.write("This line is added via append mode\n")
    f.write("This is another line added\n")
```

```python
# Read the file contents
with open("new.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()

# Print all the lines
print(lines)
```

```python
type(lines)
```

```python
for line in lines:
    print(line)
```

```python
for line in lines:
    print(line.strip())
```
