# 3. Polymorphism in Classes
different classes but same functions

```python
class India:
    def capital(self):
        print("New Delhi is capital of India")

    def language(self):
        print("Hindi is widely spoken in India")


class USA:
    def capital(self):
        print("Washington D.C. is capital of USA")

    def language(self):
        print("English is widely spoken in France")


class France:
    def capital(self):
        print("Paris is capital of USA")

    def language(self):
        print("French is widely spoken in France")
```

```python
c1 = India()
print(type(c1))
c1.capital()
c1.language()
```

```python
c2 = USA()
print(type(c2))
c2.capital()
c2.language()
```

```python
c3 = France()
print(type(c3))
c3.capital()
c3.language()
```

```python
countries = [India(), USA(), France()]

for c in countries:
    print(type(c))
    c.capital()
    c.language()
    print("\n" + "=" * 50 + "\n")
```

# Implenting polymorphism for different shapes

```python
from dataclasses import dataclass
import math


@dataclass
class Rectangle:
    width: int | float
    height: int | float

    def perimeter(self) -> int | float:
        return 2 * (self.width + self.height)

    def area(self) -> int | float:
        return self.width * self.height
```

```python
math.pi
```

```python
@dataclass
class Circle:
    radius: int | float

    def perimeter(self) -> float:
        return 2 * math.pi * self.radius

    def area(self) -> float:
        return math.pi * (self.radius**2)


from dataclasses import dataclass
import math


@dataclass
class Triangle:
    x: int | float
    y: int | float
    z: int | float

    def __post_init__(self):
        if (self.x <= 0) or (self.y <= 0) or (self.z <= 0):
            raise ValueError("Sides of triangle cannot be negative or 0")
        if not self.is_triangle():
            raise ValueError("Sum of two sides should be greater than third")

    def is_triangle(self) -> bool:
        return (
            (self.x + self.y > self.z)
            and (self.y + self.z > self.x)
            and (self.x + self.z > self.y)
        )

    def perimeter(self) -> int | float:
        return self.x + self.y + self.z

    def area(self) -> float:
        s = self.perimeter() / 2
        a = math.sqrt(s * (s - self.x) * (s - self.y) * (s - self.z))
        return a
```

```python
t1 = Triangle(x=3, y=4, z=5)
print(t1)
```

```python
shapes = [
    Triangle(x=3, y=4, z=5),
    Rectangle(width=15, height=10),
    Circle(radius=21),
    Rectangle(width=100, height=80),
    Circle(radius=40),
    Triangle(x=12, y=13, z=14),
]

for s in shapes:
    print(s)
    print(f"Perimeter : {s.perimeter():.2f}")
    print(f"Area : {s.area():.2f}")
    print("\n" + "=" * 50 + "\n")
```

# Polymorphism example from data analyst point of view

```python
import numpy as np


class NumericalSummary:

    def __init__(self, data):
        self.data = data

    def summary(self):
        return {
            "mean": np.mean(self.data),
            "max": np.max(self.data),
            "min": np.min(self.data),
            "std": np.std(self.data),
        }


class CategorialSummary:

    def __init__(self, data):
        self.data = data

    def summary(self):
        unique_vals, count = np.unique(self.data, return_counts=True)
        return dict(zip(unique_vals, count))


# Numerical data
marks = [85, 93, 86, 72, 63, 78, 95]
a = NumericalSummary(data=marks)
print(a.summary())
```

```python
temp = (12.3, 11.5, 17.1, 23, 25)
b = NumericalSummary(data=temp)
print(b.summary())
```

```python
grades = ["A", "B", "A", "B", "C", "C", "B", "F", "A", "B"]
c = CategorialSummary(data=grades)
print(c.summary())
```
